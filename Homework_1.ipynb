## Q7. Sum of weights


```python
df_asia = df[df.origin == "Asia"]
```

```python
df_asia.head()
```

```python
X = df_asia[["vehicle_weight", "model_year"]].head(7).values
```

```python
X.shape
```

```python
XTX = X.T.dot(X)
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv.dot(X.T).dot(y)
```

```python
w
```

```python
w.sum()
```
